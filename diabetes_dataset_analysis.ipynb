# Diabetes Dataset Analysis

## Step 0: Set Up
Imports

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import requests
import zipfile
import io

sns.set_theme(style="darkgrid")
sns.set_palette("husl")
```

Loading data and inspecting it

```python
url = "https://www.kaggle.com/api/v1/datasets/download/priyamchoksi/100000-diabetes-clinical-dataset"
response = requests.get(url)

# 2. Extract and read the CSV file directly from memory
with zipfile.ZipFile(io.BytesIO(response.content)) as z:
  # Find the CSV file inside the zip
  csv_filename = [f for f in z.namelist() if f.endswith(".csv")][0]

  # Read it into pandas
  with z.open(csv_filename) as f:
    diabetes_df = pd.read_csv(f)
print("----First 5 rows----")
print(diabetes_df.head(5))
print()
print("----Column Types and non-null counts----")
print(diabetes_df.info())
print()
print("----Summary statistics----")
print(diabetes_df.describe())
print()
print("----Duplicates----")
print(diabetes_df.duplicated().sum())
print()
```
